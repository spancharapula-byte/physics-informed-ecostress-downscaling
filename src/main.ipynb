```python
"""
main.py

Final entry point for the completed baseline U-Net experiment.

This script verifies and displays the outputs produced by:
    train.py
    evaluate.py
    visualization.py
"""

from pathlib import Path

import numpy as np


# ============================================================
# Paths
# ============================================================

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/ecostress_model_outputs"
)

VISUALIZATION_DIR = (
    OUTPUT_DIR / "visualizations"
)

MODEL_PATH = (
    OUTPUT_DIR / "best_unet_model.pth"
)

METRICS_PATH = (
    OUTPUT_DIR / "evaluation_metrics.npy"
)

PREDICTIONS_PATH = (
    OUTPUT_DIR / "test_predictions.npy"
)

TARGETS_PATH = (
    OUTPUT_DIR / "test_targets.npy"
)

MASK_PATH = (
    OUTPUT_DIR / "test_valid_mask.npy"
)

HISTORY_PATH = (
    OUTPUT_DIR / "training_history.npy"
)


# ============================================================
# Verify outputs
# ============================================================

def main():

    print("=" * 50)
    print("Baseline U-Net LST Downscaling Results")
    print("=" * 50)

    required_files = {
        "Trained model": MODEL_PATH,
        "Evaluation metrics": METRICS_PATH,
        "Predictions": PREDICTIONS_PATH,
        "Targets": TARGETS_PATH,
        "Valid mask": MASK_PATH,
        "Training history": HISTORY_PATH
    }

    print("\nChecking output files:")

    missing_files = []

    for name, path in required_files.items():

        exists = path.exists()

        print(
            f"{name}:",
            "FOUND" if exists else "MISSING"
        )

        if not exists:
            missing_files.append(path)

    if missing_files:

        raise FileNotFoundError(
            "Some pipeline outputs are missing:\n"
            + "\n".join(str(path) for path in missing_files)
        )


    # --------------------------------------------------------
    # Load evaluation results
    # --------------------------------------------------------

    metrics = np.load(
        METRICS_PATH,
        allow_pickle=True
    ).item()

    predictions = np.load(
        PREDICTIONS_PATH
    )

    targets = np.load(
        TARGETS_PATH
    )

    valid_mask = np.load(
        MASK_PATH
    )


    # --------------------------------------------------------
    # Display results
    # --------------------------------------------------------

    print("\nEvaluation results:")
    print(f"RMSE: {float(metrics['RMSE']):.4f} K")
    print(f"MAE:  {float(metrics['MAE']):.4f} K")
    print(f"R²:   {float(metrics['R2']):.4f}")

    print("\nSaved array shapes:")
    print("Predictions:", predictions.shape)
    print("Targets:", targets.shape)
    print("Valid mask:", valid_mask.shape)

    print("\nVisualizations:")

    comparison_path = (
        VISUALIZATION_DIR
        / "lst_comparison_patch_0.png"
    )

    history_plot_path = (
        VISUALIZATION_DIR
        / "training_history.png"
    )

    print("Comparison figure:", comparison_path)
    print("Training history:", history_plot_path)

    print("\nPipeline outputs verified successfully.")


# ============================================================
# Main
# ============================================================

if __name__ == "__main__":
    main()
```
