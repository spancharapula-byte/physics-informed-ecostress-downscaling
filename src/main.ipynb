```python
"""
main.py

Main entry point for the U-Net PINN LST downscaling project.

Pipeline:

    1. Prepare data
    2. Save processed data
    3. Train U-Net PINN
    4. Evaluate model
    5. Generate predictions
    6. Create visualizations
"""

from pathlib import Path

import numpy as np

from data_preprocessing import (
    prepare_data,
    save_processed_data
)

from train import train_model
from evaluate import evaluate_model
from inference import run_inference
from visualization import create_visualizations


# ============================================================
# Paths
# ============================================================

DATA_DIR = Path(
    "/content/drive/MyDrive/SISTER/data"
)

PREDICTION_DIR = DATA_DIR / "predictions"


# ============================================================
# Main pipeline
# ============================================================

def main():

    print(
        "\n"
        "========================================"
    )

    print(
        "U-Net PINN LST Downscaling Pipeline"
    )

    print(
        "========================================"
    )


    # ========================================================
    # 1. Data preprocessing
    # ========================================================

    print(
        "\n[1/6] Preparing data..."
    )

    X, Y, K = prepare_data()

    save_processed_data(
        X,
        Y,
        K
    )

    print(
        "Data preprocessing complete."
    )


    # ========================================================
    # 2. Train model
    # ========================================================

    print(
        "\n[2/6] Training U-Net PINN..."
    )

    train_model()

    print(
        "Training complete."
    )


    # ========================================================
    # 3. Evaluate model
    # ========================================================

    print(
        "\n[3/6] Evaluating model..."
    )

    (
        test_predictions,
        test_targets,
        valid_mask,
        metrics
    ) = evaluate_model()

    print(
        "Evaluation complete."
    )

    print(
        "\nTest Metrics:"
    )

    print(
        f"RMSE: {metrics['RMSE']:.4f}"
    )

    print(
        f"MAE:  {metrics['MAE']:.4f}"
    )

    print(
        f"R²:   {metrics['R2']:.4f}"
    )


    # ========================================================
    # 4. Generate predictions
    # ========================================================

    print(
        "\n[4/6] Generating predictions..."
    )

    run_inference()

    print(
        "Inference complete."
    )


    # ========================================================
    # 5. Prepare test predictions for visualization
    # ========================================================

    print(
        "\n[5/6] Preparing visualization data..."
    )

    PREDICTION_DIR.mkdir(
        parents=True,
        exist_ok=True
    )

    # visualization.py expects:
    #
    # predictions/predicted_lst.npy
    #
    # For the comparison plot, save the test predictions
    # so they match models/test_targets.npy.

    test_prediction_path = (
        PREDICTION_DIR
        / "predicted_lst.npy"
    )

    np.save(
        test_prediction_path,
        test_predictions
    )

    print(
        "Test predictions saved for visualization:"
    )

    print(
        test_prediction_path
    )


    # ========================================================
    # 6. Create visualizations
    # ========================================================

    print(
        "\n[6/6] Creating visualizations..."
    )

    create_visualizations()

    print(
        "\n========================================"
    )

    print(
        "Pipeline complete."
    )

    print(
        "========================================"
    )


# ============================================================
# Main
# ============================================================

if __name__ == "__main__":

    main()
```
